## 3. Müügitulu analüüs kaupluste kaupa

### Eesmärk

Uurida, kui palju müügitulu teenisid UrbanStyle'i erinevad kauplused näidisandmete põhjal.

Kasutan Pandase `groupby()` funktsiooni, et grupeerida müügitehingud asukoha järgi, ning `sum()` funktsiooni, et arvutada iga asukoha kogu müügitulu.

### Analüüsi küsimus

Milline müügikoht teenis kõige rohkem müügitulu?

### Kasutatavad veerud

- `store_location` – kaupluse asukoht
- `total_price` – tehingu summa eurodes

````python
# ÜLESANNE 3: Müügitulu analüüs kaupluste kaupa

# Grupeerime tehingud asukoha järgi ja liidame müügisummad kokku
revenue_by_store = df.groupby('store_location')['total_price'].sum()

# Sorteerime tulemused suuremast väiksemani
revenue_by_store = revenue_by_store.sort_values(ascending=False)

# Kuvame iga müügikoha kogu müügitulu
print("=== MÜÜGITULU KAUPLUSTE KAUPA ===")
print(revenue_by_store)
````

## Osa 2 – Andmete töötlemine: filtreerimine ja teisendamine

### 2.1 Connection – SQL WHERE ja Pandase boolean indexing

**Töövihik:** Nädal 7, lk 11

**Eesmärk:** Mõista, kuidas SQL-i `WHERE` tingimused vastavad Pandase DataFrame'i filtreerimisele.

SQL-is:

```sql
SELECT *
FROM sales
WHERE total_price > 100;
```

Pandases:

```python
df[df['total_price'] > 100]
```

**Selgitus:**

- `df` – meie UrbanStyle'i andmestik.
- `df['total_price']` – valime müügisumma veeru.
- `> 100` – kontrollime, millised summad ületavad 100 eurot.
- `df[...]` – tagastame ainult tingimusele vastavad read.

**Õppetund:** SQL-i `WHERE` ja Pandase boolean indexing täidavad sama eesmärki – valivad andmestikust tingimusele vastavad read.

### 2.2 – Andmete töötlemise operatsioonid
#### Samm 1. Boolean indexing (filtreerimine)

**Eesmärk:** Õppida filtreerima Pandase DataFrame'i SQL-i `WHERE` tingimuste abil tuttava loogika järgi.

**Kolm filtreerimise viisi:**

1. Lihtne tingimus – valime ainult Tallinna tellimused.
2. Mitu tingimust (`AND`) – valime Tallinna tellimused, mille summa ületab 100 eurot.
3. Välistav tingimus (`NOT`) – valime kõik tellimused, mis ei ole Tallinnast.

**Oluline süntaks:**
- `==` tähendab võrdlemist.
- `&` tähendab `AND`.
- `|` tähendab `OR`.
- `~` tähendab `NOT`.
- Mitme tingimuse korral kasutame iga tingimuse ümber sulge.

````python
# 1. Lihtne filter – Tallinna tellimused
tallinna_tellimused = df[df['store_location'] == 'Tallinn']

# 2. Mitu tingimust – üle 100 € Tallinna tellimused
suured_tallinn = df[
    (df['total_price'] > 100) &
    (df['store_location'] == 'Tallinn')
]

# 3. NOT – kõik tellimused väljaspool Tallinna
mitte_tallinn = df[~(df['store_location'] == 'Tallinn')]

# Kuvame tulemused
print("=== Tallinna tellimused ===")
display(tallinna_tellimused)

print("=== Tallinna tellimused üle 100 € ===")
display(suured_tallinn)

print("=== Tellimused väljaspool Tallinna ===")
display(mitte_tallinn)
````

|---|---:|
| `store_location == 'Tallinn'` | 9 |
| `total_price > 100` JA `store_location == 'Tallinn'` | 3 |
| `store_location != 'Tallinn'` | 11 |

#### Tulemuste tõlgendamine

**1. Lihtne filter**

Tallinnast pärineb 9 tellimust ehk 45% kõigist 20 näidistehingust.

**2. Kahe tingimusega filter (AND)**

Tallinnas oli 3 tellimust, mille summa ületas 100 eurot.

Nende tellimuste summad olid 120 €, 145 € ja 230 €, kokku 495 €.

See näitab, kuidas mitme tingimuse abil saab uurida konkreetset müügisegmenti.

**3. Välistav filter (NOT)**

Väljaspool Tallinna oli 11 tellimust: Tartus 7 ja Pärnus 4.

Tallinna 9 ja teiste linnade 11 tellimust annavad kokku 20 tehingut, mis kinnitab filtreerimise tulemuste kooskõla algandmestikuga.

#### Õppetund

- `==` kontrollib väärtuste võrdsust.
- `&` ühendab tingimused loogilise JA abil.
- `|` ühendab tingimused loogilise VÕI abil.
- `~` pöörab tingimuse vastupidiseks.
- Mitme tingimuse puhul tuleb kasutada ümarsulge.

#### Järeldus

Pandase boolean indexing võimaldab teha SQL-i `WHERE`-tingimustega sarnaseid filtreerimisi.

### 2.2 – Enne sammu 5: tingimuslik avaldis ja lambda

**Töövihik:** Nädal 7, lk 12–13

**Eesmärk:** Mõista Pythoni tingimuslikku avaldist ja `lambda` funktsiooni ning nende seost SQL-i `CASE WHEN` lausega.

#### Tingimuslik avaldis

Pythoni tingimusliku avaldise üldkuju:

`A if tingimus else B`

See tähendab:
- Kui tingimus on tõene, tagastatakse A.
- Kui tingimus on väär, tagastatakse B.

Näide:

`'Suur' if 150 > 100 else 'Väike'`

Tulemus on `'Suur'`.

#### Mis on lambda?

`lambda` on lühike, nimetu funktsioon.

Näiteks:

`lambda x: x * 2`

See võtab sisendiks `x` ja tagastab selle kahekordse väärtuse.

Tingimusliku avaldise ja lambda ühendamisel saame iga sisendväärtuse põhjal otsustada, milline tulemus tagastada.

**Seos SQL-iga:** `CASE WHEN` ja Pythoni tingimuslik avaldis võimaldavad väärtusi tingimuse alusel liigitada.

````python
# Tingimuslik avaldis ja lambda

skoor = lambda hind: 'Suur' if hind > 100 else 'Väike'

print(skoor(150))  # Suur
print(skoor(50))   # Väike
````

### Tingimuslik avaldis ja lambda – tulemused ja järeldused

**Tulemused:**
- `skoor(150)` → Suur
- `skoor(50)` → Väike

**Tõlgendus:**

Lambda-funktsioon kontrollib, kas sisestatud hind on suurem kui 100.

Kui tingimus on tõene, tagastatakse `Suur`. Vastasel juhul tagastatakse `Väike`.

**Õppetund:**

Õppisin, et `lambda` võimaldab luua lühikese funktsiooni ning `if ... else` abil saab väärtusi tingimuse põhjal liigitada.

See sarnaneb SQL-i `CASE WHEN` loogikaga.

**Järeldus:**

Tingimuslikku avaldist saab kasutada näiteks müügitehingute jagamiseks suurteks ja väikesteks tellimusteks.

   - tellimuste arv.
3. Sorteerida kliendid kogukulutuse järgi kahanevalt.
4. Lisada `vip_status` veerg.

#### VIP-staatuse tingimus

````python
# HARJUTUS 2B – KLIENTIDE OSTUKÄITUMINE

# 1. Arvutame iga kliendi ostustatistika
customer_summary = df.groupby('customer_id')['total_price'].agg(
    kogukulutus='sum',
    keskmine='mean',
    tellimusi='count'
).reset_index()

# 2. Sorteerime kogukulutuse järgi kahanevalt
customer_summary = customer_summary.sort_values(
    'kogukulutus',
    ascending=False
)

# 3. Lisame VIP-staatuse
customer_summary['vip_status'] = customer_summary['kogukulutus'].apply(
    lambda x: 'VIP' if x > 200 else 'Tavaline'
)

print(customer_summary)
````

3. Kontrollida andmestiku suurust enne ja pärast ühendamist.

#### Miks kasutada LEFT JOIN-i?

`how='left'` säilitab kõik müügitabeli read ka siis, kui mõnele müügitehingule ei leita klienditabelist vastet.

`how='inner'` jätaks alles ainult need tehingud, mille kliendi ID esineb mõlemas tabelis.

#### Äriline eesmärk

Siduda müügitulemused konkreetsete klientidega, et Marko näeks klientide nimesid, mitte ainult ID-numbreid.

````python
# HARJUTUS 2C – MERGE JA ANALÜÜS

import pandas as pd

# 1. Kliendiandmed – töövihiku näidis
customers = pd.DataFrame({
    'customer_id': [1001, 1002, 1003, 1004, 1005, 1006, 1007],
    'first_name': ['Jüri', 'Kati', 'Maris', 'Peeter', 'Liina', 'Andres', 'Tiina'],
    'email': [
        'juri@mail.ee', 'kati@mail.ee', 'maris@mail.ee',
        'peeter@mail.ee', 'liina@mail.ee',
        'andres@mail.ee', 'tiina@mail.ee'
    ]
})

# 2. Kontrollime andmestike suurust enne ühendamist
print("=== ENNE ÜHENDAMIST ===")
print("Müügitehingud:", df.shape)
print("Kliendiandmed:", customers.shape)

# 3. Ühendame müügitehingud ja kliendiandmed
merged = pd.merge(df, customers, on='customer_id', how='left')

# 4. Kontrollime tulemust
print("\n=== PÄRAST ÜHENDAMIST ===")
print("Ühendatud andmestik:", merged.shape)
display(merged.head())

# 5. TOP 5 klienti nime ja kogukulutuse järgi
top5_customers = (
    merged.groupby(['customer_id', 'first_name'])['total_price']
    .sum()
    .reset_index(name='kogukulutus')
    .sort_values('kogukulutus', ascending=False)
    .head(5)
)

print("\n=== TOP 5 KLIENTI ===")
display(top5_customers)
````

````python
import pandas as pd
import plotly.express as px

# Veendu, et sale_date on datetime tüüpi
df['sale_date'] = pd.to_datetime(df['sale_date'])

# Viitekuupäev (viimase tellimuse järgmine päev)
today = pd.to_datetime('2024-08-01')

print("Andmestiku periood:", df['sale_date'].min(), "kuni", df['sale_date'].max())
print("Viitekuupäev:", today)
````

````python
# RFM – klientide ostukäitumise analüüs

rfm = df.groupby('customer_id').agg(
    Recency=('sale_date', lambda x: (today - x.max()).days),
    Frequency=('sale_date', 'count'),
    Monetary=('total_price', 'sum')
).reset_index()

print(rfm)
````

````python
# RFM skoorid

rfm['R_score'] = pd.qcut(
    rfm['Recency'].rank(method='first'),
    q=3,
    labels=[3, 2, 1]
).astype(int)

rfm['F_score'] = pd.qcut(
    rfm['Frequency'].rank(method='first'),
    q=3,
    labels=[1, 2, 3]
).astype(int)

rfm['M_score'] = pd.qcut(
    rfm['Monetary'].rank(method='first'),
    q=3,
    labels=[1, 2, 3]
).astype(int)

rfm['RFM_score'] = (
    rfm['R_score'] +
    rfm['F_score'] +
    rfm['M_score']
)

print(rfm)
````

````python
# RFM klientide segmenteerimine

def segment(score):
    if score >= 8:
        return 'Parimad kliendid'
    elif score >= 5:
        return 'Püsikliendid'
    else:
        return 'Riskikliendid'

rfm['Segment'] = rfm['RFM_score'].apply(segment)

print(rfm[['customer_id', 'RFM_score', 'Segment']])
````

````python
# RFM segmentide visualiseerimine

segment_counts = rfm['Segment'].value_counts().reset_index()
segment_counts.columns = ['Segment', 'Klientide arv']

fig = px.pie(
    segment_counts,
    names='Segment',
    values='Klientide arv',
    title='UrbanStyle: RFM kliendisegmendid'
)

fig.show()
````

**Eesmärk:**

Analüüsida UrbanStyle'i klientide ostukäitumist RFM-meetodi abil ning jagada kliendid segmentidesse.

**Tulemused:**

- Parimad kliendid: 2 (28,6%).
- Püsikliendid: 3 (42,9%).

## RFM-kliendisegmenteerimine – tulemused ja järeldused

### Tulemused

### Järeldused

- **Recency:** aeg viimasest ostust.
- **Frequency:** ostude sagedus.
- **Monetary:** kulutatud rahasumma.

Kõige suurema rühma moodustavad püsikliendid (42,9%).

Parimatele klientidele võiks pakkuda erisoodustusi, püsiklientidele lojaalsusprogramme ning riskiklientidele personaalseid pakkumisi.

### Kokkuvõte

Õppisin kasutama Pandast klientide ostukäitumise analüüsimiseks, RFM-skooride arvutamiseks ja kliendisegmentide loomiseks.

Plotly Expressi abil visualiseerisin kliendisegmentide jaotust sektordiagrammil.

````python
import pandas as pd
import plotly.express as px

# Kontrollime, kas müügiandmed on olemas
if 'df' not in globals():
    raise ValueError("df puudub. Käivita esmalt müügiandmete loomise lahter.")

df['sale_date'] = pd.to_datetime(df['sale_date'])
today = pd.to_datetime('2024-08-01')

# 1. RFM arvutamine
recency = df.groupby('customer_id')['sale_date'].max().reset_index()
recency.columns = ['customer_id', 'last_purchase']
recency['recency_days'] = (today - recency['last_purchase']).dt.days

frequency = df.groupby('customer_id').size().reset_index(name='frequency')

monetary = df.groupby('customer_id')['total_price'].sum().reset_index()
monetary.columns = ['customer_id', 'monetary']

rfm = recency[['customer_id', 'recency_days']].merge(
    frequency, on='customer_id'
).merge(
    monetary, on='customer_id'
)

# 2. RFM-skoorid (1–5)
rfm['R_score'] = pd.qcut(
    rfm['recency_days'], q=5, labels=[5, 4, 3, 2, 1]
).astype(int)

rfm['F_score'] = pd.qcut(
    rfm['frequency'].rank(method='first'),
    q=5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm['M_score'] = pd.qcut(
    rfm['monetary'], q=5, labels=[1, 2, 3, 4, 5]
).astype(int)

rfm['RFM_score'] = (
    rfm['R_score'] + rfm['F_score'] + rfm['M_score']
)

# 3. Segmenteerimine
def assign_segment(score):
    if score >= 13:
        return 'VIP Champions'
    elif score >= 10:
        return 'Loyal Customers'
    elif score >= 7:
        return 'Potential Loyalists'
    else:
        return 'At Risk'

rfm['segment'] = rfm['RFM_score'].apply(assign_segment)

display(rfm.sort_values('RFM_score', ascending=False))

# 4. Tulpdiagramm
segment_counts = rfm['segment'].value_counts().reset_index()
segment_counts.columns = ['segment', 'count']

fig1 = px.bar(
    segment_counts,
    x='segment',
    y='count',
    title='UrbanStyle: Kliendisegmentide jaotus (RFM)',
    labels={'segment': 'Segment', 'count': 'Klientide arv'},
    color='segment'
)
fig1.show()

# 5. Hajuvusdiagramm
fig2 = px.scatter(
    rfm,
    x='recency_days',
    y='monetary',
    color='segment',
    size='frequency',
    hover_data=['customer_id'],
    title='UrbanStyle: Recency vs Monetary (RFM)',
    labels={
        'recency_days': 'Päevi viimasest ostust',
        'monetary': 'Kogukulutus (EUR)'
    }
)
fig2.show()
````

````python
vip_tulu = rfm.loc[rfm['segment'] == 'VIP Champions', 'monetary'].sum()
kogutulu = rfm['monetary'].sum()

print(f"VIP-klientide osakaal kogutuludest: {vip_tulu / kogutulu * 100:.2f}%")
````
