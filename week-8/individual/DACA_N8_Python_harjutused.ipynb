```python
# NÄDAL 8 – OSA 2.2: FUNKTSIOONI ANATOOMIA
# def alustab funktsiooni defineerimist.
# parameeter1 on kohustuslik sisend.
# parameeter2 vaikeväärtus on 10, kui teist arvu ei anta.

def funktsiooni_nimi(parameeter1, parameeter2=10):
    """Liidab kaks arvu ja tagastab tulemuse."""

    # Arvutame kahe parameetri summa.
    tulemus = parameeter1 + parameeter2

    # return annab arvutatud tulemuse funktsioonist välja.
    return tulemus
```

Harjutus 2A: Shu – kirjuta oma esimene funktsioon

Eesmärk: Luua parameetritega raportifunktsioon, mida saab korduvalt kasutada.

```python
# NÄDAL 8 – OSA 2.3: HARJUTUS 2A
# Impordime datetime, et vajadusel kasutada tänast kuupäeva.
from datetime import datetime

# Loome funktsiooni, mis koostab müügiandmetest kokkuvõtte.
# report_date=None tähendab, et kuupäev ei ole kohustuslik sisend.
def weekly_sales_report(df, report_date=None):
    """Genereeri iganädalane müügiraport.

    Args:
        df: DataFrame müügitellimustega
        report_date: Raporti kuupäev (vaikimisi täna)
    Returns:
        dict: Raporti kokkuvõte
    """

    # Kui kuupäeva pole antud, kasutame tänast kuupäeva.
    if report_date is None:
        report_date = datetime.now().strftime('%Y-%m-%d')

    # Tagastame sõnastiku, mis sisaldab raporti näitajaid.
    return {
        'report_date': report_date,
        'total_orders': len(df),
        'total_revenue': round(df['total_price'].sum(), 2),
        'avg_order': round(df['total_price'].mean(), 2),
    }
```

OSA 2.3 – Harjutus 2B: Ha – automatiseeri RFM arvutamine

Eesmärk: Luua funktsioon calculate_rfm(), mis arvutab klientide RFM-näitajad ja määrab neile segmendid.

Funktsiooni sisendid: Tellimuste DataFrame ja viitekuupäev.

Funktsiooni väljund: RFM DataFrame, mis sisaldab iga kliendi ostukäitumise näitajaid, RFM-skoore ja segmenti.

```python
# NÄDAL 8 – OSA 2.3: HARJUTUS 2B
# Automatiseerime klientide RFM-analüüsi.

import pandas as pd

def calculate_rfm(df, reference_date=None):
    """Arvuta RFM skoorid ja segmendid iga kliendi kohta."""

    # Kasutame andmete koopiat, et algset tabelit mitte muuta.
    df = df.copy()

    # Määrame kuupäeva, mille suhtes ostude värskust arvutame.
    if reference_date is None:
        reference_date = pd.to_datetime("today")
    else:
        reference_date = pd.to_datetime(reference_date)

    # Teisendame ostukuupäevad kuupäevatüübiks.
    df["sale_date"] = pd.to_datetime(df["sale_date"])

    # R – Recency: mitu päeva on viimasest ostust möödunud.
    recency = df.groupby("customer_id")["sale_date"].max().reset_index()
    recency.columns = ["customer_id", "last_purchase"]
    recency["recency_days"] = (
        reference_date - recency["last_purchase"]
    ).dt.days

    # F – Frequency: mitu ostu iga klient on teinud.
    frequency = (
        df.groupby("customer_id")
        .size()
        .reset_index(name="frequency")
    )

    # M – Monetary: kui palju iga klient kokku kulutas.
    monetary = (
        df.groupby("customer_id")["total_price"]
        .sum()
        .reset_index()
    )
    monetary.columns = ["customer_id", "monetary"]

    # Ühendame R, F ja M näitajad üheks tabeliks.
    rfm = recency[["customer_id", "recency_days"]].merge(
        frequency, on="customer_id"
    ).merge(
        monetary, on="customer_id"
    )

    # Määrame RFM-skoorid kolmes rühmas.
    rfm["R_score"] = pd.qcut(
        rfm["recency_days"].rank(method="first"),
        q=3,
        labels=[3, 2, 1]
    ).astype(int)

    rfm["F_score"] = pd.qcut(
        rfm["frequency"].rank(method="first"),
        q=3,
        labels=[1, 2, 3]
    ).astype(int)

    rfm["M_score"] = pd.qcut(
        rfm["monetary"].rank(method="first"),
        q=3,
        labels=[1, 2, 3]
    ).astype(int)

    # Liidame kolm skoori kokku.
    rfm["RFM_score"] = (
        rfm["R_score"] +
        rfm["F_score"] +
        rfm["M_score"]
    )

    # Määrame kliendisegmendi koguskoori põhjal.
    def assign_segment(score):
        if score >= 8:
            return "Tippkliendid"
        elif score >= 6:
            return "Püsikliendid"
        elif score >= 4:
            return "Potentsiaalsed kliendid"
        else:
            return "Riskikliendid"

    rfm["segment"] = rfm["RFM_score"].apply(assign_segment)

    return rfm


# Käivitame funktsiooni olemasolevate näidisandmetega.
# Viitekuupäev vastab töövihiku näitele.
rfm_result = calculate_rfm(
    sales_df,
    reference_date="2025-01-01"
)

# Kuvame kliendid RFM-skoori järgi kahanevas järjekorras.
print(rfm_result.sort_values("RFM_score", ascending=False))
```

```python
# NÄDAL 8 – OSA 2.3: HARJUTUS 2C
# Võrdleme kahe linna müügitulemusi.

def city_comparison(sales, customers, cities=None):
    """Tagastab valitud linnade tellimuste arvu ja kogukäibe."""

    # Kui linnu pole ette antud, võrdleme Tallinna ja Tartut.
    if cities is None:
        cities = ["Tallinn", "Tartu"]

    # Ühendame müügitellimused klientide linnadega.
    data = sales.merge(
        customers[["customer_id", "city"]],
        on="customer_id",
        how="left"
    )

    # Jätame alles ainult valitud linnade tellimused.
    selected = data[data["city"].isin(cities)]

    # Arvutame iga linna tellimuste arvu ja kogukäibe.
    report = selected.groupby("city").agg(
        orders=("sale_id", "count"),
        revenue=("total_price", "sum")
    ).reset_index()

    # Tagastame raporti DataFrame'ina.
    return report


# TEST 1: Tallinn ja Tartu (vaikeväärtused).
print("TEST 1 – Tallinn ja Tartu")
print(city_comparison(sales_df, customers_df))

# TEST 2: Tallinn ja Pärnu.
print("\nTEST 2 – Tallinn ja Pärnu")
print(city_comparison(
    sales_df,
    customers_df,
    cities=["Tallinn", "Parnu"]
))
```

```python
# NÄDAL 8 – HARJUTUS 3B
# Arvutame klientide RFM-näitajad ja määrame segmendid.

def transform_rfm(df, reference_date="2024-08-01"):
    """Arvuta RFM segmendid."""
    print("[TRANSFORM-RFM] Arvutan...")

    # Määrame analüüsi viitekuupäeva.
    ref = pd.to_datetime(reference_date)

    # Arvutame iga kliendi viimase ostu, ostude arvu ja kogukulutuse.
    rfm = df.groupby("customer_id").agg(
        last_purchase=("sale_date", "max"),
        frequency=("sale_date", "count"),
        monetary=("total_price", "sum")
    ).reset_index()

    # Arvutame päevade arvu viimasest ostust.
    rfm["recency_days"] = (ref - rfm["last_purchase"]).dt.days

    # Määrame kliendisegmendid töövihiku reeglite järgi.
    rfm["segment"] = rfm.apply(
        lambda row: "Tippklient"
        if row["monetary"] > 200 and row["frequency"] >= 3
        else "Püsiklient"
        if row["frequency"] >= 3
        else "Ohus"
        if row["recency_days"] > 120
        else "Tavaklient",
        axis=1
    )

    return rfm


# Kasutame harjutuse 3A tellimuste andmeid.
rfm = transform_rfm(df)

# Kuvame tulemused.
print(rfm.to_string(index=False))
```

Harjutus 3C – Iganädalase müügiaruande pipeline

Pipeline nimi: UrbanStyle'i iganädalane müügiaruanne

Sagedus: Iga nädal

EXTRACT – andmete hankimine

Allikas: Supabase'i andmebaas API kaudu

Andmed: sales ja customers tabelid

TRANSFORM – andmete töötlemine

Samm 1: Ühendan müügi- ja kliendiandmed customer_id alusel.

Samm 2: Arvutan tellimuste arvu, kogukäibe ja müügitulemused linnade kaupa.

LOAD – tulemuste salvestamine

Salvestan müügiaruande CSV-failina.

Loon Plotly graafiku ja salvestan selle HTML-failina.

```python
import pandas as pd
import plotly.express as px
from datetime import datetime

# ============================================================
# MARKO IGANADALANE RFM PIPELINE
# ============================================================

# --- EXTRACT ---
def extract():
    """Too andmed (asenda Supabase API-ga, kui ühendus olemas)."""
    print("[EXTRACT] Alustan...")

    orders = pd.DataFrame({
        'customer_id': [
            1001,1002,1003,1001,1002,1004,1003,1001,1005,1004,
            1002,1003,1005,1001,1006,1004,1002,1007,1003,1005
        ],
        'sale_date': pd.date_range('2024-01-15', periods=20, freq='10D'),
        'total_price': [
            89.99,45.50,120.00,67.30,55.00,210.00,33.50,145.00,
            78.00,92.00,160.00,44.00,88.50,230.00,37.00,175.00,
            110.00,65.00,95.00,125.00
        ],
        'store_location': [
            'Tallinn','Tartu','Tallinn','Tallinn','Tartu','Parnu',
            'Tallinn','Tallinn','Tartu','Parnu','Tartu','Tallinn',
            'Tartu','Tallinn','Parnu','Parnu','Tartu','Tallinn',
            'Tallinn','Tartu'
        ]
    })

    customers = pd.DataFrame({
        'customer_id': [1001,1002,1003,1004,1005,1006,1007],
        'first_name': ['Juri','Kati','Maris','Peeter','Liina','Andres','Tiina'],
        'last_name': ['Tamm','Kask','Sepp','Rebane','Ots','Puu','Kuusk']
    })

    print(f"[EXTRACT] {len(orders)} tellimust, {len(customers)} klienti")
    return orders, customers


# --- TRANSFORM ---
def transform(orders, customers, reference_date='2024-08-01'):
    """Puhasta, arvuta RFM, loo kuuraport."""
    print("[TRANSFORM] Alustan...")

    ref = pd.to_datetime(reference_date)
    df = pd.merge(orders, customers, on='customer_id', how='left')

    # Koostame kuupõhise müügiaruande.
    monthly = df.groupby(df['sale_date'].dt.to_period('M')).agg(
        tellimusi=('sale_date', 'count'),
        kaive=('total_price', 'sum')
    ).reset_index()

    monthly['sale_date'] = monthly['sale_date'].astype(str)

    # Arvutame klientide RFM-näitajad.
    # Segmendid põhinevad töövihiku lihtsustatud reeglitel.
    rfm = df.groupby('customer_id').agg(
        last_purchase=('sale_date', 'max'),
        frequency=('sale_date', 'count'),
        monetary=('total_price', 'sum'),
        nimi=('first_name', 'first')
    ).reset_index()

    rfm['recency_days'] = (ref - rfm['last_purchase']).dt.days

    # Tippkliendi lävend on selles harjutuses 300 eurot.
    rfm['segment'] = rfm.apply(
        lambda r: 'Tippklient'
        if r['monetary'] > 300 and r['frequency'] >= 3
        else 'Püsiklient'
        if r['frequency'] >= 3
        else 'Ohus'
        if r['recency_days'] > 120
        else 'Tavaklient',
        axis=1
    )

    print(f"[TRANSFORM] {len(rfm)} klienti segmenteeritud")
    return {'monthly': monthly, 'rfm': rfm}


# --- VALIDATE ---
def validate(results):
    """Kontrolli andmete kvaliteeti."""
    ok = (
        len(results['rfm']) > 0
        and results['monthly']['kaive'].sum() > 0
    )

    print(f"[VALIDATE] {'OK' if ok else 'PROBLEEM!'}")
    return ok


# --- LOAD ---
def load(results):
    """Salvesta CSV ja graafikud."""
    ts = datetime.now().strftime('%Y%m%d')

    # Salvestame RFM-andmed CSV-failina.
    results['rfm'].to_csv(f'rfm_report_{ts}.csv', index=False)

    # Salvestame RFM hajuvusdiagrammi HTML-failina.
    px.scatter(
        results['rfm'],
        x='recency_days',
        y='monetary',
        color='segment',
        size='frequency',
        hover_data=['nimi'],
        title=f'UrbanStyle RFM ({ts})',
        labels={
            'recency_days': 'Paevi viimasest ostust',
            'monetary': 'Kogukulutus (EUR)'
        }
    ).write_html(f'rfm_chart_{ts}.html')

    # Salvestame kuupõhise müügigraafiku HTML-failina.
    px.bar(
        results['monthly'],
        x='sale_date',
        y='kaive',
        title=f'Kuukäive ({ts})',
        labels={'sale_date': 'Kuu', 'kaive': 'EUR'}
    ).write_html(f'monthly_chart_{ts}.html')

    print("[LOAD] 1 CSV + 2 HTML salvestatud")


# --- RUN ---
print("=" * 50)
print("  MARKO IGANADALANE RFM PIPELINE")
print("=" * 50)

start = datetime.now()

orders, customers = extract()
results = transform(orders, customers)

if validate(results):
    load(results)

print(f"  VALMIS ({(datetime.now() - start).total_seconds():.1f}s)")
print("=" * 50)

# Kuvame klientide segmentide kokkuvõtte.
print("\n--- SEGMENDID ---")
print(
    results['rfm'][['nimi', 'frequency', 'monetary', 'segment']]
    .sort_values('monetary', ascending=False)
    .to_string(index=False)
)
```
